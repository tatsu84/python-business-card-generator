```python
from reportlab.pdfgen import canvas
from reportlab.pdfbase import pdfmetrics
from reportlab.pdfbase.ttfonts import TTFont
from reportlab.lib.colors import HexColor
from reportlab.lib.units import mm
import os

# ① フォントを登録
font_path = r"C:\Users\tatsu\Downloads\IPAexfont00301\IPAexfont00301\ipaexm.ttf"
pdfmetrics.registerFont(TTFont("IPAexMincho", font_path))

# ② 名刺に載せる情報を入力
company = input("会社名を入力してください：")
position = input("役職を入力してください：")
name = input("氏名を入力してください：")
postal_code = input("郵便番号を入力してください：")
address = input("住所を入力してください：")
phone = input("電話番号を入力してください：")

# ③ PDFの保存先を設定
output_path = os.path.join(
    os.path.expanduser("~"),
    "Downloads",
    "meishi_auto.pdf"
)

# ④ 名刺サイズを設定（91mm × 55mm）
width = 91 * mm
height = 55 * mm

c = canvas.Canvas(output_path, pagesize=(width, height))

# ⑤ 色を設定
navy = HexColor("#1D2B3A")
gold = HexColor("#B79A63")
dark_gray = HexColor("#343434")
light_gray = HexColor("#777777")

# ⑥ 背景
c.setFillColor(HexColor("#FFFFFF"))
c.rect(0, 0, width, height, fill=1, stroke=0)

# 左側のネイビーのライン
c.setFillColor(navy)
c.rect(0, 0, 3 * mm, height, fill=1, stroke=0)

# ⑦ 会社名
c.setFillColor(navy)
c.setFont("IPAexMincho", 10)
c.drawString(10 * mm, 46 * mm, company)

# ⑧ ゴールドのライン
c.setStrokeColor(gold)
c.setLineWidth(0.6)
c.line(10 * mm, 43 * mm, 81 * mm, 43 * mm)

# ⑨ 役職
c.setFillColor(light_gray)
c.setFont("IPAexMincho", 8)
c.drawString(10 * mm, 34 * mm, position)

# ⑩ 氏名
c.setFillColor(dark_gray)
c.setFont("IPAexMincho", 19)
c.drawString(10 * mm, 25 * mm, name)

# ⑪ 住所・電話番号
c.setFillColor(light_gray)
c.setFont("IPAexMincho", 7)
c.drawString(10 * mm, 14 * mm, postal_code)
c.drawString(10 * mm, 10 * mm, address)
c.drawString(10 * mm, 6 * mm, "TEL  " + phone)

# ⑫ PDFを保存
c.save()

print("\n名刺PDFを作成しました！")
print("保存先：", output_path)
```
